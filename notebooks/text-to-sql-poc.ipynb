```python
import os
import requests
from sqlalchemy import create_engine
from langchain_community.utilities import SQLDatabase

# 1. Download the Chinook SQLite file
db_url = "https://github.com/lerocha/chinook-database/raw/master/ChinookDatabase/DataSources/Chinook_Sqlite.sqlite"
db_path = "Chinook.db"

if not os.path.exists(db_path):
    response = requests.get(db_url)
    with open(db_path, "wb") as f:
        f.write(response.content)
    print("Chinook database downloaded!")

# 2. Connect via SQLAlchemy & LangChain
engine = create_engine(f"sqlite:///{db_path}")
db = SQLDatabase(engine)

# 3. Verify the tables
print("Tables found:", db.get_usable_table_names())
```

```python
from langchain_core.documents import Document
```

```python
# 1. Get schemas for all tables and add to Documents
table_names = db.get_usable_table_names()
documents = []

for table in table_names:
    # Get the DDL and sample rows for this specific table
    table_info = db.get_table_info(table_names=[table])
    documents.append(Document(page_content=table_info, metadata={"name": table}))
```

```python
# 1. Initialize the embedding model (requires OPENAI_API_KEY in environment)
embeddings = OpenAIEmbeddings(model="text-embedding-3-small")

# 2. Create the vector store from your existing 'documents' list
vector_db = SKLearnVectorStore.from_documents(
    documents=documents, 
    embedding=embeddings
)

print("Vector embeddings created and stored in-memory!")
```

```python
#HOW TO DO SIMILARITY SEARCH
# Define the user's question
query = "How many tracks are there in each genre?"

# Perform the search (k=2 means return the top 2 most relevant tables)
related_docs = vector_db.similarity_search(query, k=2)

# Display the results
for doc in related_docs:
    print(f"Relevant Table found: {doc.metadata['name']}")
    # print(doc.page_content) # Uncomment to see the full DDL
```

```python
# INTEGRATION | USER QUERY --> VECTOR STORE --> LLM
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

# 1. Initialize the LLM (GPT-4o or GPT-3.5-turbo)
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

# 2. Define the User Question
user_question = "Which artists have the most rock tracks?"

# 3. STEP 1: VECTOR SEARCH (Find relevant tables)
# We search the vector_db for tables related to the question
relevant_docs = vector_db.similarity_search(user_question, k=3)
dynamic_schema = "\n\n".join([doc.page_content for doc in relevant_docs])

# 4. STEP 2: PROMPT ENGINEERING
# We create a prompt that injects the dynamic schema found by the vector db
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a SQL expert. Use the provided schema to write a valid SQLite query."),
    ("human", f"Schema:\n{dynamic_schema}\n\nQuestion: {user_question}\nSQL Query:")
])

# 5. STEP 3: EXECUTION
# Combine the prompt and LLM to get the SQL string
sql_generator = prompt | llm

generated_sql = sql_generator.invoke({})
print(f"--- Generated SQL ---\n{generated_sql.content}")
```

```python
#CREATE STRUCTURED LLM FOR PARSING RESPONSE
from pydantic import BaseModel, Field

# 1. Define your desired structure
class SQLResponse(BaseModel):
    sql: str = Field(description="The generated SQLite query")
    explanation: str = Field(description="A brief explanation of the logic")
```

```python
# 1. Keep your SQLResponse class as is
# 2. Bind it to the LLM
structured_llm = llm.with_structured_output(SQLResponse)

# 3. Simplify your prompt (no format instructions needed)
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a SQL expert. Use the provided schema to write a valid SQLite query."),
    ("human", "Schema:\n{schema}\n\nQuestion: {question}")
])

# 4. Create the chain
structured_chain = prompt | structured_llm

# 5. Invoke (notice we don't pass format_instructions anymore)
response = structured_chain.invoke({
    "schema": dynamic_schema,
    "question": user_question
})

# Access data directly as an object
print(f"Generated SQL: {response.sql}")
print(f"Explanation: {response.explanation}")
```

```python
# 1. Execute the SQL using the LangChain database wrapper
# query_results = db.run(response.sql)

# 2. Print the raw results
print("--- Query Results ---")
df_tracks = pd.read_sql(response.sql, engine)

# Display the formatted table in your notebook
df_tracks.head()
```

```python
#SELF CORRECTION IF QUERY FAILS 
# 1. Attempt execution
max_retries = 2
current_query = response.sql

for i in range(max_retries):
    try:
        df_results = pd.read_sql(current_query, engine)
        print("✅ Query executed successfully!")
        display(df_results)
        break
    except Exception as e:
        print(f"❌ Attempt {i+1} failed. Asking LLM to fix...")
        
        # 2. Prepare a "Fix-it" prompt
        fix_prompt = f"""
        The following SQL query failed:
        {current_query}
        
        Error Message: {str(e)}
        
        Please provide the corrected SQL query based on the error above.
        """
        
        # 3. Get the corrected response
        correction = structured_chain.invoke({
            "schema": dynamic_schema, 
            "question": fix_prompt
        })
        current_query = correction.sql
```
